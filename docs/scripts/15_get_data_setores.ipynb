## Detalhe

Fazer o modo _batch_


````python
r, dd = setores.detalhe(id_setor=4146)

dd
````

Explorar 

````python
# BS4
soup = BeautifulSoup(r.text, "html.parser")
pprint.pprint(soup.prettify())
````

````python
# 1. Extrair o Nome do Setor
nome_setor = soup.find("h2", class_="setortitle").get_text(strip=True)

# 2. Extrair os pares Chave/Valor da lista de dados
dados = {}
for dl in soup.find_all("dl", class_="dl-lista-telefonica"):
    chave = dl.find("dt").get_text(strip=True)
    # Rmove o ponto e vírgula e espaços em branco do final
    valor = dl.find("dd").get_text(strip=True).rstrip(";").strip()
    dados[chave] = valor

# 3. Extrair o Código do Imóvel (caso precise)
codigo_imovel = soup.find("input", id="hdCodigoImovel")["value"]

# Exibir os resultados
print(f"Setor: {nome_setor}")
print(f"Código Imóvel: {codigo_imovel}")
print("\nDetalhes:")
for chave, valor in dados.items():
    print(f"  {chave}: {valor}")
````
