## Detalhes

É possível consultar detalhes de um imóvel


````python
id_imovel = 291
````

````python
imovel.detalhe(id_imovel=id_imovel)
````

<br>

O método `obter_imovel` reflete um _endpoint_ do TJSP que é público. Contudo, ao que tudo indicada, ele retorna o mesmo que o método `detalhe`.


````python
imovel.obter_imovel(id_imovel=id_imovel)
````

<br>

E é possível fazer isso em modo _batch_, ou seja, pesquisando **em lote**. Para esse método eu, inicialmente, usava parte do nome das unidades (que obtive juntamente com os municípios), para pesquisar. Observei que nem todos os imóveis eram mapeados com esse método, dai "chumbei" no código a iteração entre 0 e 1200.


````python
# Detalhes dos Imóveis
imovel.detalhe_batch()

# Results
imovel.df_detalhes.info()
imovel.df_detalhes.head()
````

<br>

Com isso chegamos à 321 comarcas. Acredito que o erro da Vila Mimosa esteja também aqui.


````python
len(set(imovel.df_detalhes['comarca_tjsp']))
````

<br>

Não está!


````python
mask = imovel.df_detalhes['comarca_tjsp'].str.contains('Vila')
imovel.df_detalhes[mask]
````

<br>

Salvamos o resultado.


````python
imovel.df_detalhes.to_csv(
    path_or_buf=output_path_tab / "imovel.csv",
    index=False,
)
````
