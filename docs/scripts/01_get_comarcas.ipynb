# RAJs, Circunscrição Judiciária e Comarcas

Por meio do _site_ [Quem Somos](https://www.tjsp.jus.br/QuemSomos/QuemSomos/RegioesAdministrativasJudiciarias) obtenho as informações das _Regiões Administrativas Judiciárias_.


```python
from tjsp_unidades import pagina
from tjsp_unidades.paths import output_path_tab
```

```python
quem_somos = pagina.QuemSomos()
```

<br>

---

## RAJs

Pego os dados das RAJs


```python
df_raj = quem_somos.rajs

df_raj.head()
```

```python
filename = "RAJs"

df_raj.to_csv(
    path_or_buf=output_path_tab / f"{filename}.csv",
    index=False,
)

df_raj.to_excel(
    excel_writer=output_path_tab / f"{filename}.xlsx",
    index=False,
    sheet_name=f"{filename}",
)
```

<br>

---

## Circunscrições Judiciárias (CJs)

Obtenhos também os dados de _Circunscrições Judiciárias_.


```python
df_cj = quem_somos.cjs

df_cj.head()
```

```python
filename = 'Circunscrições Judiciárias'

df_cj.to_csv(
    path_or_buf=output_path_tab / f'{filename}.csv',
    index=False,
)

df_cj.to_excel(
    excel_writer=output_path_tab / f'{filename}.xlsx',
    sheet_name=f'{filename}',
    index=False,
)
```

<br>

---

## Comarcas

Por fim, obtemos a lista de Comarcas por CJ.


```python
df_comarca = quem_somos.comarcas

df_comarca.head()
```

```python
filename = "Comarcas e CJs"

df_comarca.to_csv(
    path_or_buf=output_path_tab / f"{filename}.csv",
    index=False,
)

df_comarca.to_excel(
    excel_writer=output_path_tab / f"{filename}.xlsx",
    index=False,
    sheet_name=f"{filename}",
)
```
