## Lista de Municípios

Inicialmente instanciamos a classe `Municipio`. Com essa classe é possível reproduzir o comportamento do _site_ do TJSP, ou seja, ao "inputar" um termo com mais de 3 caracteres, a API "devolve" uma lista de 10 municípios prováveis.


```python
municipio = extract.Municipio()
```

<br>

Criamos função que pega o Nome e Código do município, de acordo com o TJSP.


```python
# Apenas Exemplo da Função!
municipio.search(termo='San')
```

<br>

```python
municipio._lista_municipios[0:5]
```
